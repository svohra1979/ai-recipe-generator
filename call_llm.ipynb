```python
from dotenv import load_dotenv
load_dotenv(override=True)
```

```python
from google import genai

client = genai.Client()

interaction = client.interactions.create(
    model="gemini-3.8-flash",
    input="Explain how AI works in a few words"
)

print(interaction.output_text)
```

```python
def generate_recipe(ingredients,cuisine,diet,goal,weight):

    '''
    generate a recipe using the ingredients provided, in accordance to the cuisine that has been mentioned.
    The diet can be classified in "vegan","veg","carnivore","any", create the recipe accordingly.
    Depending on the "goal" and "weight" provided, for example "gaining","cutting","maintainence", alter the recipe's calories.
    '''
    
    prompt= f'''
    generate a recipe using the ingredients provided, in accordance to the cuisine that has been mentioned.
    The diet can be classified in "vegan","veg","carnivore","any", create the recipe accordingly.
    Depending on the "goal" and "weight" provided, for example "gaining","cutting","maintainence", alter the recipe's calories.
    goal : {goal}
    weight : {weight}
    diet :{diet}
    ingredients: :{','.join(ingredients)}
    cuisine :{cuisine}
    the recipe should not exceed 100 words
    '''
    
    interaction = client.interactions.create(
        model="gemini-3.8-flash",
        input=prompt
    )
    
    print(interaction.output_text)
```

```python
ingredients=["chicken","basil","spices"]
generate_recipe(ingredients,"italian","any","maintainence",75)
```
